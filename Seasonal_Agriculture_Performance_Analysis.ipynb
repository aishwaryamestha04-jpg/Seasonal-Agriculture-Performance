```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from scipy import stats

import warnings
warnings.filterwarnings('ignore')

print("Libraries imported successfully!")
```

```python
df = pd.read_csv("seasonal_agriculture_performance_dataset.csv")

print("Dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
missing = df.isnull().sum()

print(missing[missing > 0])
```

```python
df['Rainfall_mm'] = df['Rainfall_mm'].fillna(df['Rainfall_mm'].median())

df['Soil_Moisture_pct'] = df['Soil_Moisture_pct'].fillna(
    df['Soil_Moisture_pct'].median()
)

df['Yield_Tonnes_Ha'] = df['Yield_Tonnes_Ha'].fillna(
    df['Yield_Tonnes_Ha'].median()
)
```

```python
print(df.isnull().sum().sum())
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
print(df['Season'].unique())
```

```python
df['Season'].value_counts()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(data=df, x='Season')

plt.title('Number of Farms by Season')
plt.xlabel('Season')
plt.ylabel('Number of Farms')

plt.show()
```

```python
season_yield = df.groupby('Season')['Yield_Tonnes_Ha'].mean().sort_values(
    ascending=False
)

print(season_yield)
```

```python
plt.figure(figsize=(8,5))

season_yield.plot(kind='bar')

plt.title('Average Crop Yield by Season')
plt.xlabel('Season')
plt.ylabel('Average Yield (Tonnes/Ha)')

plt.xticks(rotation=0)
plt.show()
```

```python
season_production = df.groupby('Season')['Production_Tonnes'].mean()

print(season_production)
```

```python
season_profit = df.groupby('Season')['Profit_INR'].mean().sort_values(
    ascending=False
)

print(season_profit)
```

```python
plt.figure(figsize=(8,5))

season_profit.plot(kind='bar')

plt.title('Average Profit by Season')
plt.xlabel('Season')
plt.ylabel('Average Profit (INR)')

plt.xticks(rotation=0)
plt.show()
```

```python
season_revenue = df.groupby('Season')['Revenue_INR'].mean()

print(season_revenue)
```

```python
plt.figure(figsize=(8,5))

season_revenue.plot(kind='bar')

plt.title('Average Revenue by Season')
plt.xlabel('Season')
plt.ylabel('Average Revenue (INR)')

plt.xticks(rotation=0)
plt.show()
```

```python
season_cost = df.groupby('Season')['Total_Cost_INR'].mean()

print(season_cost)
```
